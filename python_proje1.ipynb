```python
ad_soyad=input("Ad Soyad giriniz:")#kullanıcıdan ad soyad bilgileri alındı(input her zaman metin döndürur)
aylik_ucret=float(input("Aylık ücret giriniz:"))#kullanıcıdan aylik ücret alindi floata çevrildi
sadakat_ayi=int(input("kaç aydır müşteri:"))#kullanıcıdan kaç aydir müşteri oldugu bilgisi alindi int türüne çevrildi
aktif=input("müşteri aktif mi(e/h):")#kullanıcıdan müşterinin aktif olup olmadigi bilgisi alindi
if aktif.lower()=="e":#lower ile alinan bilgi küçük harf olucak
  aktif_mi=True#alınan bilgi e ise true
else:
  aktif_mi=False

hizmetler=[]#boş bir liste oluşturulur hizmetler burda tutulucak
for id in range(5):#döngü 5 kere dönücek
   hizmet=input(f"{id+1}.hizmeti giriniz:")#1.hizmeti giriniz vs diye hizmetler istenir
   hizmetler.append(hizmet)#append listeye eleman ekler

#sözlük oluşturulup kullanıcıdan alinan bilgiler listeleniyo
musteri={#sözlük oluşturma
    "ad_soyad":ad_soyad,
    "ayliK_ucret":aylik_ucret,
    "sadakat_ayi":sadakat_ayi,
    "aktif_mi":aktif_mi,
    "hizmetler":hizmetler
}
if aylik_ucret>500 or sadakat_ayi>24:#şartlardan bir tanesi bile dogruysa VIP olur
  print("VIP müşteri indirim tanımlandı")
else:
  print("indirim tanımlanmadı")

import random#rastgele sayi üretme
print("BÜYÜK İSİM",ad_soyad.upper())#isim büyük harflere çevrilir
customerID= "IST-2026-"+str(random.randint(1000,9999))#bu sayilar arasinda rastgele sayi üretilir ve sayi yaziya çevrilir
print("müşteri ID",customerID)

print("müsteri bilgileri",musteri)
```
